```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data = {
    "House_Size":[600,800,1000,1200,1400,1600,1800,2000,2200,2400],
    "House_Price":[25,32,38,45,52,58,65,72,80,87]
}
df = pd.DataFrame(data)
print("Dataset: ")
print(df)
```

```python
plt.scatter(df["House_Size"],df['House_Price'])
plt.xlabel("House Size (sq.ft.)")
plt.ylabel("House Price (Lakhs)")
plt.title("House Size Vs House Price")
plt.grid(True)
plt.show()
```

```python
#Observation: The Scatter Plot shows a strong positive linear relationship between House Size and House Price. As the House size
# increases the house price also increases. The data points are closely aligned along an upward trend, indicating that Simple Linear Regression is an
# Appropriate model for predicting house prices based on house size
```

```python
train = df.iloc[:8]
test = df.iloc[8:]
X_train = train["House_Size"].values
Y_train = train["House_Price"].values
X_test = test["House_Size"].values
Y_test = test["House_Price"].values
```

```python
mean_x = np.mean(X_train)
mean_y = np.mean(Y_train)

num = np.sum((X_train - mean_x)*(Y_train - mean_y))
den = np.sum((X_train - mean_x)**2)
theta1 = num / den #Slope
theta0 = mean_y - theta1 * mean_x #Intercept
print("\n Regression Constant: ",theta0)
print("Regression Coefficient: ",theta1)

```

```python
Y_pred = theta0 + theta1 * X_test
print("\n Actual vs Predicted")
for actual, pred in zip(Y_test,Y_pred):
    print(f"Actual = {actual},Predicted = {pred:.2f}")
    
```

```python
sizes = [1500,1700,2500]
print("\nPredictions:")
for s in sizes:
    price = theta0 + theta1 * s
    print(f"{s} sq.ft.  --> {price:.2f} Lakhs")
    
```

```python
plt.scatter(df["House_Size"],df["House_Price"], label = "Data")
x = np.array(df['House_Size'])
y = theta0 + theta1 * x
plt.plot(x,y,color='red',label = 'Regression Line')
plt.xlabel("House Size")
plt.ylabel("House Price")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Evaluation metrices
mae = np.mean(np.abs(Y_test - Y_pred))
mse = np.mean((Y_test - Y_pred)**2)
rmse = np.sqrt(mse)

ss_total = np.sum((Y_test - np.mean(Y_test))**2)
ss_res = np.sum((Y_test - Y_pred)**2)
r2 = 1 - (ss_res / ss_total)

print("\n Evaluation Metrices")
print("MAE: ",mae)
print("MSE: ",mse)
print("RMSE: ",rmse)
print("R2 Score: ",r2)
```
